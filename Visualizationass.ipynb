```python
##1.CountPlot
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")
plt.rcParams['figure.figsize']=(10,10)

dataset = sns.load_dataset("titanic")

sns.countplot(data = dataset, x = "embarked", hue ="sex")

plt.title("Passenger Count by Embarked Port")
plt.xlabel("Embarked Port")
plt.ylabel("Passenger Count")

plt.legend(title="Sex")
plt.show()
print(dataset["embarked"].value_counts())
```

Here in the countplot embarkrd port "S" has the highest number of passenger boarding and next comes "C" and the "Q". More male passegers had boarded at the ports than the female passngers.

```python
##2.Swarmplot
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")
plt.rcParams['figure.figsize']=(10,10)

dataset = sns.load_dataset("titanic")

#dataset = dataset.dropna(how="any")

sns.swarmplot(data = dataset, x="pclass", y="age", hue="sex")

plt.title("Age Distribution Across Passenger Classes by Sex")
plt.xlabel("Passenger Class")
plt.ylabel("Age")

plt.legend(title="Sex")
plt.show()
```

Most of the passengers were young adults with ages varying across 3 different classes and third class has younger passengers compared with the other two.

```python
##3.Piechart
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")
plt.rcParams['figure.figsize']=(10,10)

dataset = sns.load_dataset("titanic")

gender_count = dataset["sex"].value_counts()

plt.pie(gender_count,labels = gender_count.index, autopct = '%.1f%%',explode =[0,0.1],startangle = 140, shadow = True)  

plt.title("Gender Distribution of Titanic Passengers")
plt.axis('equal')
plt.show()

```

The male passengers has the largest percentage than the female passengers.

```python
##4.HeatMap
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")
plt.rcParams['figure.figsize']=(10,10)

dataset = sns.load_dataset("titanic")

data = dataset[["age", "fare", "pclass", "sibsp", "parch", "survived"]]

correlation_matrix = data.corr()

plt.figure(figsize=(8,4))
sns.heatmap(correlation_matrix, annot = True,  cmap = 'coolwarm')
plt.title("Correlation Matrix of Titanic Dataset")

plt.show()
```

Fare and passenger class (pclass) showed a negative correlation, indicating that higher-class passengers generally paid higher fares.Passenger class and survival also showed a negative correlation.

```python
##5.Violin Plot
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")
plt.rcParams['figure.figsize']=(10,10)

dataset = sns.load_dataset("titanic")

plt.figure(figsize=(8,4))
sns.catplot(data = dataset, x ="pclass", y ="fare",kind = "violin", hue ="sex")

plt.title("Fare Distribution by Class and Gender")
plt.xlabel("Passenger Class")
plt.ylabel("Fare")

plt.show()
```

First-class passengers generally paid higher fares than the other two classes ( second and third classes)

```python
##6.Scatter Plot
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")
plt.rcParams['figure.figsize']=(10,10)

dataset = sns.load_dataset("titanic")

plt.figure(figsize=(8,4))
sns.relplot(data = dataset, x ="age", y ="fare",kind = "scatter", hue ="survived");

plt.title("Age vs Fare by Survival Status")
plt.xlabel("Age (Years)")
plt.ylabel("Fare")

plt.legend(title="Survival", labels=["Did Not Survive", "Survived"])

plt.show()
```

Most passengers paid lower fares,but few paid higher fares. Survivors and non-survivors appeared across different ages and fare amounts with no relation with age and fare.
